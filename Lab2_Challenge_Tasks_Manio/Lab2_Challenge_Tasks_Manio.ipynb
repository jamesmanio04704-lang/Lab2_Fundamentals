```python

# [EXERCISE 1] SENSOR MONITORING SYSTEM
SURNAME = "MANIO"
SEED_NUM = 3  # Replace with your actual SEED_NUM from Task 4

# 1. Generate unique sensor data based on surname length and SEED_NUM
raw_readings = [f"{ord(SURNAME[0])}", "invalid_str", f"{SEED_NUM * 15}", "200", f"{SEED_NUM * 5 - 10}"]
operating_limit = (SEED_NUM * 10) + 20

valid_count = 0
invalid_count = 0
results = []

print("=== SENSOR DATA PROCESSING LOG ===")
for entry in raw_readings:
    try:
        val = int(entry)
        if val < 0:
            status = "INVALID (Negative Reading)"
            invalid_count += 1
        elif val > operating_limit:
            status = "CRITICAL (Over Limit)"
            valid_count += 1
        else:
            status = "NOMINAL (Within Range)"
            valid_count += 1
        results.append((entry, status))
        print(f"Reading '{entry}': {status}")
    except ValueError:
        invalid_count += 1
        results.append((entry, "INVALID (Non-Numeric)"))
        print(f"Reading '{entry}': INVALID (Non-Numeric)")

print("\n=== SUMMARY REPORT ===")
print(f"Total Readings Processed: {len(raw_readings)}")
print(f"Valid Readings: {valid_count} | Invalid Readings: {invalid_count}")
```

```python
# [EXERCISE 2] SIGNAL DIAGNOSTIC SYSTEM
SURNAME = "MANIO"
SEED_NUM = 3

# 1. Generate raw signal string using SURNAME and SEED_NUM
raw_signal = f"  Sig-{SURNAME}-{SEED_NUM * 107}#!  "

# 2. Normalize signal (strip whitespace and uppercase)
processed_signal = raw_signal.strip().upper()

digits = 0
letters = 0
specials = 0

print(f"Raw Signal: '{raw_signal}'")
print(f"Processed Signal: '{processed_signal}'")
print("\n--- Character Analysis ---")

for char in processed_signal:
    if char.isalpha():
        letters += 1
        ctype = "Letter"
    elif char.isdigit():
        digits += 1
        ctype = "Digit"
    else:
        specials += 1
        ctype = "Special Symbol/Delimiter"
    print(f"Char: '{char}' -> ASCII: {ord(char)} | Type: {ctype}")

# Signal Classification based on composition
if digits > letters:
    classification = "Telemetry / Numerical Data Stream"
else:
    classification = "Command / Text Protocol"

print(f"\nSignal Classification: {classification}")
```

```python
# [EXERCISE 3] ACCESS AUTHENTICATION SYSTEM
SURNAME = "MANIO"
SEED_NUM = 3

# Generate password and attempt limit
target_password = f"{SURNAME}{SEED_NUM * 10}".upper()
attempt_limit = SEED_NUM + 2

# Simulated login attempts
simulated_inputs = ["ADMIN", "WRONG123", target_password]

attempts = 0
authenticated = False

print(f"Target Security Credential: {target_password}")
print(f"Allowed Attempts Limit: {attempt_limit}\n")

for pwd_try in simulated_inputs:
    attempts += 1
    normalized_try = pwd_try.strip().upper()
    print(f"Attempt {attempts}/{attempt_limit}: Entered '{normalized_try}'")
    
    if normalized_try == target_password:
        authenticated = True
        print(">>> ACCESS GRANTED: Security Clearance Verified.")
        break
    else:
        print(">>> ACCESS DENIED: Incorrect Credential.")
    
    if attempts >= attempt_limit:
        print(">>> SYSTEM LOCKOUT: Maximum attempt threshold reached.")
        break

final_state = "UNLOCKED" if authenticated else "LOCKED / SUSPENDED"
print(f"\nFinal System State: {final_state}")
```
